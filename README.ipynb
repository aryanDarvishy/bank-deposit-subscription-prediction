# 📊 Project 4 — Прогнозирование отклика клиента на предложение банка

## 💡 Описание проекта
Банк проводит маркетинговые кампании, цель которых — привлечь клиентов к открытию депозитов.  
Каждый контакт с клиентом (по телефону, почте, при визите и т. д.) требует времени и денег.  
Задача проекта — построить модель машинного обучения, которая сможет **предсказать, согласится ли клиент открыть депозит**, основываясь на его характеристиках и истории взаимодействий с банком.

---

## 🎯 Цель проекта
Разработать модель, которая поможет банку:
- Определить клиентов, **наиболее склонных** к открытию депозита;
- Повысить **эффективность маркетинговых кампаний**;
- Снизить **издержки** на взаимодействие с незаинтересованными клиентами.

---

## 🧠 Этапы работы

### 1. Первичная обработка данных
- Обнаружение и обработка пропусков;
- Очистка выбросов;
- Приведение типов данных к корректным форматам.

### 2. Разведывательный анализ данных (EDA)
- Анализ распределений признаков;
- Изучение зависимости между признаками и целевой переменной;
- Визуализация результатов и предварительные выводы.

### 3. Отбор и преобразование признаков
- Кодирование категориальных признаков (`LabelEncoder`, `OneHotEncoder`);
- Масштабирование числовых признаков (`MinMaxScaler`);
- Формирование финального набора признаков для обучения моделей.

### 4. Решение задачи классификации
Построены и протестированы базовые модели:
- **Логистическая регрессия**;
- **Решающее дерево**;
- Подбор гиперпараметров с помощью **GridSearchCV**.

### 5. Ансамблевые методы
- **Стекинг**: объединение логистической регрессии, решающего дерева и градиентного бустинга.  
  Метамодель — логистическая регрессия.
- **Бэггинг и бустинг** для повышения устойчивости модели.
- **Оптимизация гиперпараметров с помощью Optuna** на примере случайного леса.
